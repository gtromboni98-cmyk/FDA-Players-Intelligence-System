# 03 - Advanced Stats Analysis

This notebook focuses on fetching, cleaning, and analyzing advanced match statistics for players.

**Prerequisites**: Complete previous notebooks or run scripts up to `scripts/03_fetch_advanced_stats.py`


## Setup & Data Loading

```python
import sys
from pathlib import Path

# Add project root to path
project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

# Import modules
from src.features import DataProcessor
from src.utils import DataManager

import pandas as pd
import numpy as np
from pprint import pprint

print("✓ Modules imported successfully")
```

```python
# Initialize utilities
manager = DataManager()
processor = DataProcessor()

# Load raw match stats (from script 03 or later)
try:
    df_match_stats = manager.load_pickle('data/raw/match_stats.pkl')
    print(f"✓ Loaded match stats with shape: {df_match_stats.shape}")
except FileNotFoundError:
    print("⚠ Match stats not found. Run scripts/03_fetch_advanced_stats.py first")
    df_match_stats = None

# Try to load cleaned stats (from script 04)
try:
    df_clean = manager.load_pickle('data/processed/players_stats_clean.pkl')
    print(f"✓ Loaded cleaned stats with shape: {df_clean.shape}")
except FileNotFoundError:
    print("⚠ Cleaned stats not available yet")
    df_clean = None
```

## Step 1: Explore Advanced Stats Structure

```python
# Generate statistical summary
if df_clean is not None:
    print("✓ Statistical Summary of Advanced Stats:")
    print(df_clean.describe())
    print(f"\nTotal players: {df_clean['wyId'].nunique() if 'wyId' in df_clean.columns else 'N/A'}")
elif df_match_stats is not None:
    print("✓ Statistical Summary (uncleaned):")
    print(df_match_stats.describe())
```

## Step 3: Statistical Summary

```python
# Clean the data using DataProcessor
if df_match_stats is not None:
    print("Cleaning match stats...")
    
    # Handle missing data
    df_cleaned = processor.handle_missing_data(df_match_stats, strategy='drop')
    
    print(f"\n✓ Before cleaning: {len(df_match_stats)} rows")
    print(f"✓ After cleaning:  {len(df_cleaned)} rows")
    print(f"✓ Rows removed: {len(df_match_stats) - len(df_cleaned)}")
    
    # Save cleaned data
    manager.save_pickle(df_cleaned, 'data/processed/players_stats_clean.pkl')
    print("\n✓ Saved cleaned stats to data/processed/players_stats_clean.pkl")
else:
    if df_clean is not None:
        print("✓ Using pre-cleaned stats")
```

## Step 2: Data Cleaning

```python
# Examine raw match stats structure
if df_match_stats is not None:
    print("✓ Match Stats DataFrame:")
    print(f"  Shape: {df_match_stats.shape}")
    print(f"  Columns: {df_match_stats.columns.tolist()}")
    print(f"\nData types:")
    print(df_match_stats.dtypes)
    print(f"\nFirst few rows:")
    print(df_match_stats.head())
    print(f"\nMissing values:")
    print(df_match_stats.isnull().sum())
```
